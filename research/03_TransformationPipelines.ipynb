```python
def apply_transformation_steps(df, steps):
    """
    Apply a series of transformation steps to a dataframe.

    Parameters:
    df (pd.DataFrame): The input dataframe to be transformed.
    steps (list): A list of transformation functions to be applied.

    Returns:
    pd.DataFrame: The transformed dataframe.
    """
    for step in steps:
        df = step(df)
    return df

# Apply the transformation steps to the dataframe
transformed_df = apply_transformation_steps(df, steps)
```
